```python
import os

demo_file = '/content/drive/MyDrive/SadTalker_Patched_2026/src/gradio_demo.py'

if os.path.exists(demo_file):
    with open(demo_file, 'r') as f:
        lines = f.readlines()

    fixed_lines = []
    for line in lines:
        # We look for the lines we inserted and force them to 8 spaces
        # (This is the standard depth for a method inside a class in this repo)
        if any(x in line for x in ['import shutil', 'final_output =', 'shutil.copy', 'return final_output']):
            indent = "        " # Exactly 8 spaces
            content = line.strip()
            fixed_lines.append(f"{indent}{content}\n")
        else:
            fixed_lines.append(line)

    with open(demo_file, 'w') as f:
        f.writelines(fixed_lines)
    print("✅ Indentation Hard-Aligned: 8-space standard applied.")
else:
    print("❌ File not found. Verify Drive mount.")
```

```python
import os

demo_file = '/content/drive/MyDrive/SadTalker_Patched_2026/src/gradio_demo.py'

if os.path.exists(demo_file):
    with open(demo_file, 'r') as f:
        content = f.read()

    # 1. Add shutil to the top if it's not there
    if "import shutil" not in content[:100]:
        content = "import shutil\n" + content

    # 2. Clean up the messy indentation from the previous attempt at the bottom
    # We'll make sure the return block is clean and uses the global shutil
    lines = content.splitlines()
    final_lines = []
    for line in lines:
        if 'import shutil' in line and '        ' in line: # Remove the nested one
            continue
        final_lines.append(line)

    with open(demo_file, 'w') as f:
        f.write("\n".join(final_lines))

    print("✅ Shutil promoted to Global Scope. Logic cleaned.")
else:
    print("❌ File not found.")
```

```python
%cd /content

with open('gradio_app/main.py', 'w') as f:
    f.write("""
import gradio as gr
import os

# Function to get secret from environment (injected by Cloud Run)
def get_password():
    return os.environ.get("MY_APP_PASSWORD", "default_fallback_pass")

def gruß_funktion(name):
    return f"Hallo {name}!"

demo = gr.Interface(fn=gruß_funktion, inputs="text", outputs="text")

if __name__ == "__main__":
    demo.launch(
        server_name="0.0.0.0",
        server_port=8080,
        auth=("admin", get_password())
    )
""")
```
